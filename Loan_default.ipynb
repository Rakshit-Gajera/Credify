# Linerar-Regression and gradient descent

### Part A

```python
from sklearn.linear_model import LinearRegression
X = df[['Income']] #Feature
y = df['LoanAmount'] #Target
model = LinearRegression()
model.fit(X, y)
print("Sklearn Weight (m):", model.coef_[0])
print("Sklearn Bias (c):", model.intercept_)
```

```python
predictions = model.predict(X)
print("First 5 Predictions:")
print(predictions[:5])
```

###
LoanAmount = m × Income + c

Gradient Descent is used to find the optimal values of m and c by minimizing Mean Squared Error (MSE).

### Part B

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

X = df[['Income']].values
y = df['LoanAmount'].values

scaler = StandardScaler()
X = scaler.fit_transform(X)

w = 0
b = 0

learning_rate = 0.01
epochs = 1000

loss_history = []

n = len(X)

for i in range(epochs):

    y_pred = w * X.flatten() + b

    error = y_pred - y

    # Loss
    loss = np.mean(error ** 2)
    loss_history.append(loss)

    # Gradients
    dw = (2 / n) * np.sum(error * X.flatten())
    db = (2 / n) * np.sum(error)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Gradient Descent Weight (m):", w)
print("Gradient Descent Bias (c):", b)
```

```python
import matplotlib.pyplot as plt

plt.plot(loss_history)
plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

```python
print("----- Comparison -----")

print("Sklearn Weight:", model.coef_[0])
print("Gradient Descent Weight:", w)

print()

print("Sklearn Bias:", model.intercept_)
print("Gradient Descent Bias:", b)
```
